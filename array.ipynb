```python
import numpy as np
arr_1d=np.array([1,2,4,5])
print(arr_1d)
```

```python
arr_2d=np.array([(1,2),(3,4)])
print(arr_2d)
```

```python
arr_3d=np.zeros([2,3,4])
print(arr_3d)
print("size:",arr_3d.size)
print("shape:",arr_3d.shape)
print("dimension:",arr_3d.ndim)
```

```python
#datatype
arr_int32=np.array([1,2,3],dtype=np.int32)
print("data type of arr_int4=32:",arr_int32.dtype)
```

```python
arr_float32=np.array([1.2,3.3,5.6], dtype=np.float32)
print("memory",arr_float32.itemsize*arr_float32.size,"bytes")
```
